A tabela deverá ser dividida em 3
- homens e mulheres por região com curso superior
- homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
- homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar	

Exemplo 1

```python
# : = todas as linhas  |  0:4 = colunas 0, 1, 2 e 3 (as 4 primeiras)
formacao_total = formacao_bruto.iloc[3:36, 0:4].copy()
```

```python
formacao_total.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_total
```

```python
formacao_total["percentual_mulheres"] = ((formacao_total['total_mulheres'] / formacao_total['total']) * 100).round(2)
```

# Existe algum estado em que a média de mulheres com graduação está abaixo que homens?

```python
# Percentual de mulheres no Brasil; estados abaixo desse valor
pct_brasil = formacao_total.loc[
    formacao_total["uf_regiao"] == "Brasil", "percentual_mulheres"
].item()

regioes = ["Brasil", "Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
estados = formacao_total[~formacao_total["uf_regiao"].isin(regioes)].copy()

abaixo_brasil = estados[estados["percentual_mulheres"] < pct_brasil]

print(f"Percentual de mulheres no Brasil: {pct_brasil}%")
print(f"Estados abaixo do Brasil: {len(abaixo_brasil)}")
abaixo_brasil[["uf_regiao", "percentual_mulheres"]].sort_values("percentual_mulheres") 
```

Exemplo 2 — Ciência, Tecnologia, Engenharias e Matemática (STEM)

```python
formacao_stem = formacao_bruto.iloc[3:36, [0, 4, 5, 6]].copy()
formacao_stem.columns = ["uf_regiao", "total", "homens", "mulheres"]
for col in ["total", "homens", "mulheres"]:
    formacao_stem[col] = pd.to_numeric(formacao_stem[col], errors="coerce")
formacao_stem["percentual_mulheres"] = (
    formacao_stem["mulheres"] / formacao_stem["total"] * 100
).round(2)

pct_brasil_stem = formacao_stem.loc[
    formacao_stem["uf_regiao"] == "Brasil", "percentual_mulheres"
].item()
estados_stem = formacao_stem[~formacao_stem["uf_regiao"].isin(regioes)].copy()
abaixo_stem = estados_stem[estados_stem["percentual_mulheres"] < pct_brasil_stem]

print(f"Percentual de mulheres no Brasil (STEM): {pct_brasil_stem}%")
print(f"Estados abaixo do Brasil: {len(abaixo_stem)}")
abaixo_stem[["uf_regiao", "percentual_mulheres"]].sort_values("percentual_mulheres")
```

Exemplo 3 — Educação, Serviços pessoais, Saúde e Bem-estar

```python
formacao_saude = formacao_bruto.iloc[3:36, [0, 7, 8, 9]].copy()
formacao_saude.columns = ["uf_regiao", "total", "homens", "mulheres"]
for col in ["total", "homens", "mulheres"]:
    formacao_saude[col] = pd.to_numeric(formacao_saude[col], errors="coerce")
formacao_saude["percentual_mulheres"] = (
    formacao_saude["mulheres"] / formacao_saude["total"] * 100
).round(2)

pct_brasil_saude = formacao_saude.loc[
    formacao_saude["uf_regiao"] == "Brasil", "percentual_mulheres"
].item()
estados_saude = formacao_saude[~formacao_saude["uf_regiao"].isin(regioes)].copy()
abaixo_saude = estados_saude[estados_saude["percentual_mulheres"] < pct_brasil_saude]

print(f"Percentual de mulheres no Brasil (Educação/Saúde): {pct_brasil_saude}%")
print(f"Estados abaixo do Brasil: {len(abaixo_saude)}")
abaixo_saude[["uf_regiao", "percentual_mulheres"]].sort_values("percentual_mulheres")
```
