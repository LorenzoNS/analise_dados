```python
bruto = pd.read_excel(path, sheet_name="Municípios", header=None)
```

```python
bruto
```

```python
tabela_cortada = bruto.iloc[3:5573, 1:8].copy()
```

```python
tabela_cortada = tabela_cortada.reset_index(drop=True)
tabela_cortada.columns = [
    "UF",
    "COD. UF",
    "COD. MUNIC",
    "NOME DO MUNICÍPIO",
    "População",
    "População 2010",
    "População Censo 2022",
]
for col in tabela_cortada.columns[4:6]:
    tabela_cortada[col] = pd.to_numeric(tabela_cortada[col], errors="coerce")
tabela_cortada
```

```python
tabela_cortada["Diferença"] = tabela_cortada["População Censo 2022"] - tabela_cortada["População 2010"]
tabela_cortada.sort_values("Diferença", ascending=False).reset_index(drop=True)
```

```python
tabela_cortada.to_csv(r"H:\Python\Etapa3\Aula4\Tabela4.csv")
```

```python
pop_estado = (
    tabela_cortada.groupby(["COD. UF", "UF"], as_index=False)
      .agg(
          populacao_2010=("População 2010", "sum"),
          populacao_2022=("População Censo 2022", "sum"),
      )
)

pop_estado["Diferença"] = pop_estado["populacao_2022"] - pop_estado["populacao_2010"]

pop_estado.sort_values("Diferença", ascending=False).reset_index(drop=True)
```

```python
pop_estado.to_csv(r"H:\Python\Etapa3\Aula4\Tabela4_2.csv")
```
